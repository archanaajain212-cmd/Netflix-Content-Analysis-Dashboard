```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# Locate the project folder
project_folder = Path.cwd().parent.parent

# Find the dataset
csv_files = list((project_folder / "Dataset").glob("*.csv"))

# Load the dataset
if not csv_files:
    print("Error: No CSV file found in the Dataset folder.")
else:
    df = pd.read_csv(csv_files[0])

    print("Dataset loaded successfully!")
    print("Total rows:", df.shape[0])
    print("Total columns:", df.shape[1])

    print("\nFirst 5 rows:")
    print(df.head())
```

```python
# Dataset Overview

print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nFirst 5 Rows:")
display(df.head())
```

```python
# Missing Values Analysis

print("Missing Values Analysis")
print("=" * 40)

missing_values = df.isnull().sum()

print(missing_values)

print("\nTotal Missing Values:", df.isnull().sum().sum())
```

```python
# Duplicate Records Analysis

print("Duplicate Records Analysis")
print("=" * 40)

duplicate_records = df.duplicated().sum()

print("Total Duplicate Records:", duplicate_records)

# Display duplicate records if available
if duplicate_records > 0:
    print("\nDuplicate Records:")
    display(df[df.duplicated()].head(10))
else:
    print("No duplicate records found.")
```
